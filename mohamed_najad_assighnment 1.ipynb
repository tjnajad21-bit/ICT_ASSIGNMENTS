# Q8

```python
import math
a = float(input('enter side a '))
b = float(input('enter side b '))
c = float(input('enter side c '))
s =(a + b + c ) / 2
area = math.sqrt(s*(s-a)*(s-b)*(s-c))
print('area of the triangle: {:.2f}'.format(area))
```

# Q9 

```python
length = float(input('enter wall length'))
height = float(input('enter wall height'))
windowlength = 1.5
windowheight = 1.2
windowarea = windowlength*windowheight*2
wallarea = length*height
paintablearea = wallarea-windowarea
costpersqm = 25
totalcost = paintablearea*costpersqm
print('total area to be painted: {:.1f} sq.m'.format(paintablearea))
print('total painting cost: ₹{:.1f}'.format(totalcost))
```
